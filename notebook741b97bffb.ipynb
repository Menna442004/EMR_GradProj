```python
# Open and read the broken training file
with open("d3pm/train.py", "r") as f:
    lines = f.readlines()

# Find the broken 'else:' and fix the block
for i, line in enumerate(lines):
    if "else:" in line and i+1 < len(lines) and "Fallback if validation skips" in lines[i+1]:
        lines[i] = "" # Delete the 'else:' line entirely
        
        # Un-indent the fallback block below it (shift left by 4 spaces)
        j = i + 1
        while j < len(lines):
            # If it's indented 12 spaces, drop it down to 8 spaces
            if lines[j].startswith("            "):
                lines[j] = lines[j][4:]
            elif lines[j].strip() != "":
                break # Stop when we hit normal code again
            j += 1
        break

# Save the fixed code
with open("d3pm/train.py", "w") as f:
    f.writelines(lines)
    
print("✅ d3pm/train.py successfully fixed!")
```

```python
with open("d3pm/model.py", "r") as f:
    content = f.read()

# 1. Fix the multiplication step
content = content.replace(
    "q_xtm1_given_x0 = ac_p * F.one_hot(", 
    "q_xtm1_given_x0 = ac_p.view(-1, 1, 1, 1) * F.one_hot("
)

# 2. Fix the addition step at the end of the calculation
content = content.replace(
    ").float() + (1 - ac_p) / K", 
    ").float() + (1 - ac_p.view(-1, 1, 1, 1)) / K"
)

with open("d3pm/model.py", "w") as f:
    f.write(content)

print("✅ d3pm/model.py successfully fixed!")
```

```python
import os
import sys

# If evaluate.py is in a different folder, Python might not find it.
# We will append the current directory just to be safe!
sys.path.append(os.getcwd())
try:
    from evaluate import evaluate_detailed
except ImportError:
    print("❌ Could not import evaluate.py. Make sure it is in your current directory!")
    print("Current directory:", os.getcwd())
    sys.exit()

# 1. Map your models to their output files
# (Update these paths if your files saved somewhere else!)
input_file = "../data/example_input.txt" 
model_outputs = {
    "SeqGAN": "out.txt",
    "VAE": "out_vae.txt",
    "D3PM": "out_d3pm.txt",
    "Diffusion": "out_diffusion.txt"
}

# 2. Parse the input conditions from the dataset
conditions = []
try:
    with open(input_file, 'r') as f:
        for line in f:
            # Clean up the brackets (e.g. "[SAT] [MAR] [False] [182]" -> ["SAT", "MAR", "False", "182"])
            conds = line.replace('[', '').replace(']', '').strip().split()
            if conds:
                conditions.append(conds)
except FileNotFoundError:
    print(f"❌ Could not find {input_file}. Check your data path!")
    sys.exit()

# 3. Evaluate each model and print a formatted table
print(f"{'Model':<12} | {'Overall CSR':<12} | {'Day':<8} | {'Month':<8} | {'Leap':<8} | {'Decade':<8}")
print("-" * 70)

for model_name, file_path in model_outputs.items():
    if not os.path.exists(file_path):
        print(f"{model_name:<12} | --- File '{file_path}' not found ---")
        continue
        
    preds = []
    with open(file_path, 'r') as f:
        for line in f:
            # Grab just the generated date (usually the last string on the line)
            date_str = line.strip().split()[-1] 
            preds.append(date_str)
            
    # Safety check to ensure predictions line up with inputs
    if len(preds) != len(conditions):
        print(f"{model_name:<12} | Error: {len(preds)} predictions vs {len(conditions)} conditions")
        continue
        
    # Run your evaluate.py magic!
    results = evaluate_detailed(preds, conditions)
    
    # Extract the metrics (multiply by 100 for percentages)
    csr = results.get('all', 0) * 100
    day = results.get('day', 0) * 100
    month = results.get('month', 0) * 100
    leap = results.get('leap', 0) * 100
    decade = results.get('decade', 0) * 100
    
    print(f"{model_name:<12} | {csr:>9.2f}%   | {day:>6.1f}% | {month:>6.1f}% | {leap:>6.1f}% | {decade:>6.1f}%")

print("-" * 70)
print("Evaluation Complete!")
```
